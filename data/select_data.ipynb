```python
final_df = final_df.reset_index(drop=True)
final_df.to_json("question_bank_final.jsonl", orient="records", lines=True)
final_df
```

```python
# Filter non-train questions
non_train = final_df[final_df['dataset'] != 'train'].reset_index(drop=True)

# Define desired truth pattern (15 total, with 3 incorrects)
pattern = [True, True, False, True, False, False, True, True, False, True, False, False] # 3 fix left random; 3 : 3, 4: 4, 5: 3

# Split based on correctness
correct = non_train[non_train['is_correct'] == True].copy()
incorrect = non_train[non_train['is_correct'] == False].copy()

# Check if we have enough
assert len(correct) >= pattern.count(True), "Not enough correct samples!"
assert len(incorrect) >= pattern.count(False), "Not enough incorrect samples!"

# Build reordered list
reordered = []
correct_index = 0
incorrect_index = 0

for is_corr in pattern:
    if is_corr:
        reordered.append(correct.iloc[correct_index])
        correct_index += 1
    else:
        reordered.append(incorrect.iloc[incorrect_index])
        incorrect_index += 1

# Create new DataFrame
non_train_reordered = pd.DataFrame(reordered).reset_index(drop=True)

# Combine with training questions at the top
train_questions = final_df[final_df['dataset'] == 'train']
final_reordered_df = pd.concat([train_questions, non_train_reordered], ignore_index=True)
final_reordered_df
```

```python
final_reordered_df.to_json("question_bank_final.jsonl", orient="records", lines=True)
```
